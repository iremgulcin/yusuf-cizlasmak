```python
import matplotlib.pyplot as plt
import numpy as np

# Veri
clusters = ['Cluster 0', 'Cluster 1', 'Cluster 2', 'Cluster 3', 'Cluster 4', 'Cluster 5']
istenen = np.array([230, 464, 37, 390, 399, 28])
istenmeyen = np.array([140, 192, 304, 108, 103, 401])
filigran = np.array([142, 126, 17, 124, 82, 6])

# Figür ve axis oluştur
fig, ax = plt.subplots(figsize=(12, 6))

# Bar'ları çiz
bar_width = 0.8
bars1 = ax.bar(clusters, istenen, bar_width, label='İstenen', color='#2ecc71')
bars2 = ax.bar(clusters, istenmeyen, bar_width, bottom=istenen, label='İstenmeyen', color='#e74c3c')
bars3 = ax.bar(clusters, filigran, bar_width, bottom=istenen + istenmeyen, label='Filigran', color='#3498db')

# Değerleri bar'ların üzerine ekle
for i in range(len(clusters)):
    # İstenen değerleri
    ax.text(i, istenen[i]/2, str(istenen[i]), 
            ha='center', va='center')
    
    # İstenmeyen değerleri
    ax.text(i, istenen[i] + istenmeyen[i]/2, str(istenmeyen[i]), 
            ha='center', va='center')
    
    # Filigran değerleri
    ax.text(i, istenen[i] + istenmeyen[i] + filigran[i]/2, str(filigran[i]), 
            ha='center', va='center')

# Grafik özelliklerini ayarla
ax.set_ylabel('Sayı')
ax.set_title('Cluster Analizi')
ax.legend()


plt.show()
```
